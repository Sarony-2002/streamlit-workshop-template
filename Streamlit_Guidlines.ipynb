🎈 Chapter 5 - User Inputs

Explain interactivity.

Use:


```python
st.text_input()

st.number_input()

st.slider()

st.checkbox()

st.radio()

st.selectbox()

st.multiselect()

st.button()
```

Example:


```python
name = st.text_input("What's your name?")

age = st.slider("Age",0,100,20)

favorite = st.selectbox(
    "Favorite Language",
    ["Python","Java","C++"]
)

st.write(name)
st.write(age)
st.write(favorite)
```

Congratulations! 🎉 You've officially built something more interactive than PowerPoint.

📋 Chapter 10 - Metrics

```python
st.metric(
    "Revenue",
    "$52K",
    "+12%"
)
```

Then columns

```python
c1,c2,c3=st.columns(3)

with c1:
    st.metric("Sales","100","+5%")

with c2:
    st.metric("Users","250","+20%")

with c3:
    st.metric("Orders","54","-2%")
```

🎈 Chapter 15 - Progress

```python
st.progress()

st.spinner()

st.balloons()

st.snow()
```

Example

```python
import time

bar=st.progress(0)

for i in range(100):
    time.sleep(.02)
    bar.progress(i+1)

st.balloons()
```

🎛️ Chapter 16 - Session State

Explain why values disappear.

```python
if "count" not in st.session_state:
    st.session_state.count=0

if st.button("Click"):
    st.session_state.count+=1

st.write(st.session_state.count)
```
